# QAOA Portfolio Optimization

Bu not defteri sabit sayida varlik secen ikili Markowitz problemini QAOA ile cozer. Degerler aciklanabilir bir ornek icin sentetiktir.

```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import pandas as pd

project_root = Path.cwd()
if not (project_root / 'src').exists():
    project_root = project_root.parent
sys.path.insert(0, str(project_root / 'src'))

from quantum.portfolio_qaoa import QAOAPortfolioConfig, solve_qaoa_portfolio
```

```python
expected_returns = pd.Series({
    'AAPL': 0.12,
    'MSFT': 0.10,
    'NVDA': 0.16,
    'JNJ': 0.07,
})
covariance = pd.DataFrame(
    [[0.040, 0.018, 0.025, 0.006],
     [0.018, 0.030, 0.020, 0.005],
     [0.025, 0.020, 0.070, 0.004],
     [0.006, 0.005, 0.004, 0.020]],
    index=expected_returns.index,
    columns=expected_returns.index,
)
expected_returns.to_frame('expected_return')
```

```python
config = QAOAPortfolioConfig(budget=2, risk_aversion=0.5, reps=2, maxiter=100, seed=123)
result = solve_qaoa_portfolio(expected_returns, covariance, config)
print('Selected assets:', ', '.join(result.selected_assets))
print('Objective value:', result.objective_value)
result.weights.to_frame()
```

```python
axis = result.weights.sort_values().plot.barh(figsize=(8, 4), color='tab:blue')
axis.set_xlabel('Equal weight after binary selection')
axis.set_ylabel('Asset')
plt.tight_layout()
```
